```python
# CAPSTONE — BASELINE REFRESH OPPORTUNITY SCORE

baseline = analysis_df.copy()

baseline["baseline_selected"] = (
    (baseline["content_age_days"] >= 180) &
    (baseline["search_volume"].fillna(0) > 0)
)

baseline["baseline_score"] = (
    baseline["search_volume"]
    .fillna(0)
    .where(baseline["baseline_selected"], 0)
)

baseline["reason_code"] = "not_selected"

baseline.loc[
    baseline["baseline_selected"],
    "reason_code"
] = "old_content_with_search_demand"

baseline["action"] = "NO_ACTION"

baseline.loc[
    baseline["baseline_selected"],
    "action"
] = "REVIEW_REFRESH"

baseline_queue = (
    baseline[
        [
            "content_hash_id",
            "client_hash_id",
            "content_age_days",
            "search_volume",
            "baseline_score",
            "reason_code",
            "action"
        ]
    ]
    .sort_values(
        ["baseline_score", "search_volume"],
        ascending=False
    )
    .reset_index(drop=True)
)

print("Total rows:", len(baseline_queue))
print(
    "Selected for review:",
    (baseline_queue["baseline_score"] > 0).sum()
)
print(
    "Not selected:",
    (baseline_queue["baseline_score"] == 0).sum()
)

print("\nTOP 20 BASELINE OPPORTUNITIES")
display(baseline_queue.head(20))
```

```python
# ============================================================
# CAPSTONE — CTR OPPORTUNITY LABEL
# ============================================================

df = analysis_df.copy()

# Only pages with meaningful search visibility
visible = df[
    (df["impressions_30d"] > 0) &
    (df["avg_position_30d"] > 0)
].copy()

print("Visible pages:", len(visible))

# ------------------------------------------------------------
# 1. Position bands
# ------------------------------------------------------------

visible["position_band"] = pd.cut(
    visible["avg_position_30d"],
    bins=[0, 3, 10, 20, 50, float("inf")],
    labels=["Top 3", "4-10", "11-20", "21-50", "50+"],
    include_lowest=True
)

# ------------------------------------------------------------
# 2. Smoothed CTR
# ------------------------------------------------------------
# Avoid treating very small CTR samples as equally reliable.
# Bayesian-style smoothing using global CTR.

global_clicks = visible["clicks_30d"].sum()
global_impressions = visible["impressions_30d"].sum()

global_ctr = (
    global_clicks / global_impressions
    if global_impressions > 0
    else 0
)

SMOOTHING_IMPRESSIONS = 100

position_stats = (
    visible
    .groupby("position_band", observed=True)
    .agg(
        clicks=("clicks_30d", "sum"),
        impressions=("impressions_30d", "sum")
    )
)

position_stats["smoothed_ctr"] = (
    position_stats["clicks"] +
    global_ctr * SMOOTHING_IMPRESSIONS
) / (
    position_stats["impressions"] +
    SMOOTHING_IMPRESSIONS
)

print("\nSmoothed CTR benchmark by position:")
display(
    position_stats[["clicks", "impressions", "smoothed_ctr"]]
)

# ------------------------------------------------------------
# 3. Expected CTR benchmark
# ------------------------------------------------------------

visible["benchmark_ctr"] = (
    visible["position_band"]
    .map(position_stats["smoothed_ctr"])
    .astype(float)
)

# ------------------------------------------------------------
# 4. CTR gap
# ------------------------------------------------------------

visible["ctr_gap"] = (
    visible["benchmark_ctr"] -
    visible["ctr_30d"]
)

# ------------------------------------------------------------
# 5. Opportunity definition
# ------------------------------------------------------------

MIN_IMPRESSIONS = 100

visible["ctr_opportunity"] = (
    (visible["ctr_gap"] > 0) &
    (visible["impressions_30d"] >= MIN_IMPRESSIONS)
)

print(
    "\nCTR opportunity pages:",
    visible["ctr_opportunity"].sum()
)

print(
    "\nCTR opportunity rate:",
    round(
        visible["ctr_opportunity"].mean() * 100,
        2
    ),
    "%"
)
```

```python
# CAPSTONE — RANK CTR OPPORTUNITIES

import numpy as np
import pandas as pd

visible["opportunity_score"] = (
    visible["ctr_gap"].clip(lower=0)
    * np.log1p(visible["impressions_30d"])
)

visible["reason_code"] = np.where(
    visible["ctr_opportunity"],
    "below_position_band_ctr",
    "no_opportunity"
)

visible["recommended_action"] = np.where(
    visible["ctr_opportunity"],
    "REVIEW_TITLE_SNIPPET",
    "MONITOR"
)

ctr_queue = (
    visible[
        [
            "content_hash_id",
            "client_hash_id",
            "content_age_days",
            "impressions_30d",
            "clicks_30d",
            "ctr_30d",
            "avg_position_30d",
            "benchmark_ctr",
            "ctr_gap",
            "opportunity_score",
            "reason_code",
            "recommended_action"
        ]
    ]
    .sort_values(
        "opportunity_score",
        ascending=False
    )
    .reset_index(drop=True)
)

print("Ranked opportunity queue:", len(ctr_queue))

print("\nTOP 25 CTR OPPORTUNITIES")
display(ctr_queue.head(25))
```

```python
# ============================================================
# CAPSTONE — ML CTR OPPORTUNITY MODEL
# ============================================================

import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    classification_report,
    roc_auc_score,
    average_precision_score,
    precision_score,
    recall_score,
    f1_score
)

# ------------------------------------------------------------
# 1. Prepare modeling dataset
# ------------------------------------------------------------

model_df = visible.copy()

# Use the opportunity label created earlier
model_df = model_df[
    model_df["impressions_30d"] >= 100
].copy()

model_df["target"] = model_df["ctr_opportunity"].astype(int)

print("Modeling rows:", len(model_df))
print("\nTarget distribution:")
print(model_df["target"].value_counts())
print("\nTarget percentage:")
print(model_df["target"].mean() * 100)


# ------------------------------------------------------------
# 2. Safe predictive features
# ------------------------------------------------------------

feature_cols = [
    "keyword_char_count",
    "keyword_token_count",
    "public_url_char_count",
    "public_url_path_depth",
    "content_title_char_count",
    "content_title_token_count",
    "search_volume",
    "content_type",
    "main_intent",
    "content_age_days",
    "client_has_gsc",
    "client_has_ga4"
]

X = model_df[feature_cols].copy()
y = model_df["target"].copy()

print("\nFeatures used:")
for feature in feature_cols:
    print("-", feature)


# ------------------------------------------------------------
# 3. Identify feature types
# ------------------------------------------------------------

numeric_features = [
    "keyword_char_count",
    "keyword_token_count",
    "public_url_char_count",
    "public_url_path_depth",
    "content_title_char_count",
    "content_title_token_count",
    "search_volume",
    "content_age_days"
]

categorical_features = [
    "content_type",
    "main_intent",
    "client_has_gsc",
    "client_has_ga4"
]


# ------------------------------------------------------------
# 4. Preprocessing
# ------------------------------------------------------------

numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median"))
])


categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(
        handle_unknown="ignore",
        sparse_output=True
    ))
])

preprocessor = ColumnTransformer([
    ("numeric", numeric_pipeline, numeric_features),
    ("categorical", categorical_pipeline, categorical_features)
])


# ------------------------------------------------------------
# 5. Train / test split
# ------------------------------------------------------------

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\nTrain rows:", len(X_train))
print("Test rows:", len(X_test))


# ------------------------------------------------------------
# 6. Random Forest model
# ------------------------------------------------------------

model = RandomForestClassifier(
    n_estimators=250,
    max_depth=12,
    min_samples_leaf=20,
    class_weight="balanced",
    random_state=42,
    n_jobs=-1
)

pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("model", model)
])


# ------------------------------------------------------------
# 7. Train
# ------------------------------------------------------------

pipeline.fit(X_train, y_train)


# ------------------------------------------------------------
# 8. Predictions
# ------------------------------------------------------------

pred = pipeline.predict(X_test)
prob = pipeline.predict_proba(X_test)[:, 1]


# ------------------------------------------------------------
# 9. Model evaluation
# ------------------------------------------------------------

roc_auc = roc_auc_score(y_test, prob)
avg_precision = average_precision_score(y_test, prob)
precision = precision_score(y_test, pred, zero_division=0)
recall = recall_score(y_test, pred, zero_division=0)
f1 = f1_score(y_test, pred, zero_division=0)

print("\n================ MODEL RESULTS ================")
print("ROC-AUC:", round(roc_auc, 4))
print("Average Precision:", round(avg_precision, 4))
print("Precision:", round(precision, 4))
print("Recall:", round(recall, 4))
print("F1:", round(f1, 4))

print("\nClassification Report:")
print(classification_report(
    y_test,
    pred,
    zero_division=0
))
```

```python
# ============================================================
# CAPSTONE — BASELINE VS ML MODEL
# ============================================================

from sklearn.metrics import (
    roc_auc_score,
    average_precision_score,
    precision_score,
    recall_score,
    f1_score
)

# ------------------------------------------------------------
# 1. ML model scores
# ------------------------------------------------------------

ml_score = prob


# ------------------------------------------------------------
# 2. Simple baseline
# ------------------------------------------------------------
# Baseline assumption:
# pages with greater search demand are higher-priority
# opportunities.

baseline_score = (
    X_test["search_volume"]
    .fillna(X_train["search_volume"].median())
    .astype(float)
)


# ------------------------------------------------------------
# 3. Baseline metrics
# ------------------------------------------------------------

baseline_roc_auc = roc_auc_score(
    y_test,
    baseline_score
)

baseline_ap = average_precision_score(
    y_test,
    baseline_score
)


# ------------------------------------------------------------
# 4. ML metrics
# ------------------------------------------------------------

ml_roc_auc = roc_auc_score(
    y_test,
    ml_score
)

ml_ap = average_precision_score(
    y_test,
    ml_score
)


# ------------------------------------------------------------
# 5. Compare
# ------------------------------------------------------------

comparison = pd.DataFrame({
    "Method": [
        "Search-volume baseline",
        "Random Forest model"
    ],
    "ROC-AUC": [
        baseline_roc_auc,
        ml_roc_auc
    ],
    "Average Precision": [
        baseline_ap,
        ml_ap
    ]
})

comparison["ROC-AUC"] = comparison["ROC-AUC"].round(4)
comparison["Average Precision"] = comparison["Average Precision"].round(4)

display(comparison)


# ------------------------------------------------------------
# 6. Improvement
# ------------------------------------------------------------

roc_improvement = (
    ml_roc_auc - baseline_roc_auc
)

ap_improvement = (
    ml_ap - baseline_ap
)

print("\n================ IMPROVEMENT ================")

print(
    "ROC-AUC improvement:",
    round(roc_improvement, 4)
)

print(
    "Average Precision improvement:",
    round(ap_improvement, 4)
)
```

```python
# ============================================================
# FINAL MODEL — OUT-OF-FOLD PROBABILITIES FOR RANKING
# ============================================================

from sklearn.model_selection import StratifiedKFold, cross_val_predict

# ------------------------------------------------------------
# 1. Create final Random Forest
# ------------------------------------------------------------

final_model = RandomForestClassifier(
    n_estimators=250,
    max_depth=12,
    min_samples_leaf=20,
    class_weight="balanced",
    random_state=42,
    n_jobs=-1
)

final_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("model", final_model)
])

# ------------------------------------------------------------
# 2. Generate out-of-fold probabilities
# ------------------------------------------------------------
# Each row gets a probability from a model that did NOT
# train on that row.

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

all_prob = cross_val_predict(
    final_pipeline,
    X,
    y,
    cv=cv,
    method="predict_proba",
    n_jobs=-1
)[:, 1]

print("OOF probabilities generated:", len(all_prob))

# ------------------------------------------------------------
# 3. Fit final model on ALL data
# ------------------------------------------------------------
# This fitted model is used later for feature importance.

final_pipeline.fit(X, y)

print("Final model trained on all rows:", len(X))
```

```python
# ============================================================
# CAPSTONE — FINAL RANKED OPPORTUNITY RECOMMENDATIONS
# ============================================================

# ------------------------------------------------------------
# 1. Create recommendation dataframe
# ------------------------------------------------------------

recommendations = X.copy()

# IMPORTANT:
# Use OOF probabilities generated in Cell 23
recommendations["model_probability"] = all_prob

# ------------------------------------------------------------
# 2. Keep identifiers and business context
# ------------------------------------------------------------

recommendations["content_hash_id"] = model_df.loc[
    X.index,
    "content_hash_id"
].values

recommendations["client_hash_id"] = model_df.loc[
    X.index,
    "client_hash_id"
].values

recommendations["content_age_days"] = model_df.loc[
    X.index,
    "content_age_days"
].values

recommendations["search_volume"] = model_df.loc[
    X.index,
    "search_volume"
].values

recommendations["content_type"] = model_df.loc[
    X.index,
    "content_type"
].values

recommendations["main_intent"] = model_df.loc[
    X.index,
    "main_intent"
].values

# ------------------------------------------------------------
# 3. Opportunity tiers
# ------------------------------------------------------------

HIGH_THRESHOLD = 0.85
MEDIUM_THRESHOLD = 0.70

recommendations["reason_code"] = np.where(
    recommendations["model_probability"] >= HIGH_THRESHOLD,
    "HIGH_MODEL_OPPORTUNITY",
    np.where(
        recommendations["model_probability"] >= MEDIUM_THRESHOLD,
        "MEDIUM_MODEL_OPPORTUNITY",
        "LOW_MODEL_OPPORTUNITY"
    )
)

# ------------------------------------------------------------
# 4. Recommended action
# ------------------------------------------------------------

recommendations["recommended_action"] = np.select(
    [
        recommendations["model_probability"] >= HIGH_THRESHOLD,
        recommendations["model_probability"] >= MEDIUM_THRESHOLD
    ],
    [
        "PRIORITIZE_TITLE_SNIPPET_REVIEW",
        "REVIEW_TITLE_SNIPPET"
    ],
    default="MONITOR"
)

# ------------------------------------------------------------
# 5. Final ranked queue
# ------------------------------------------------------------

ranked_queue = recommendations[
    [
        "content_hash_id",
        "client_hash_id",
        "content_age_days",
        "search_volume",
        "content_type",
        "main_intent",
        "model_probability",
        "reason_code",
        "recommended_action"
    ]
].copy()

ranked_queue = (
    ranked_queue
    .sort_values(
        "model_probability",
        ascending=False
    )
    .reset_index(drop=True)
)

ranked_queue["rank"] = (
    ranked_queue.index + 1
)

ranked_queue = ranked_queue[
    [
        "rank",
        "content_hash_id",
        "client_hash_id",
        "content_age_days",
        "search_volume",
        "content_type",
        "main_intent",
        "model_probability",
        "reason_code",
        "recommended_action"
    ]
]

print("\nRANKED OPPORTUNITY ENGINE")
print(
    "Total ranked pages:",
    len(ranked_queue)
)

print("\nTOP 30 MODEL-RANKED OPPORTUNITIES")

display(
    ranked_queue.head(30)
)
```

```python
# ============================================================
# FINAL OPPORTUNITY SEGMENTATION
# ============================================================

ranked_queue["search_volume_clean"] = ranked_queue["search_volume"].fillna(0)

# High-confidence model threshold
HIGH_THRESHOLD = 0.85

# Minimum search demand for a validated business opportunity
MIN_SEARCH_VOLUME = 20

# 1. Validated opportunities
validated_opportunities = ranked_queue[
    (ranked_queue["model_probability"] >= HIGH_THRESHOLD) &
    (ranked_queue["search_volume_clean"] >= MIN_SEARCH_VOLUME)
].copy()

# 2. Model-only opportunities

model_only_opportunities = ranked_queue[
    (ranked_queue["model_probability"] >= HIGH_THRESHOLD) &
    (ranked_queue["search_volume_clean"] < MIN_SEARCH_VOLUME)
].copy()

print("========== FINAL OPPORTUNITY SUMMARY ==========")

print("Total ranked pages:", len(ranked_queue))

print(
    "Validated opportunities:",
    len(validated_opportunities)
)

print(
    "Model-only opportunities:",
    len(model_only_opportunities)
)

print(
    "Total high-probability opportunities:",
    len(validated_opportunities) + len(model_only_opportunities)
)

print("\nValidated opportunities:")
display(
    validated_opportunities[
        [
            "rank",
            "content_hash_id",
            "client_hash_id",
            "content_age_days",
            "search_volume",
            "content_type",
            "main_intent",
            "model_probability",
            "reason_code",
            "recommended_action"
        ]
    ].head(25)
)

print("\nModel-only opportunities:")
display(
    model_only_opportunities[
        [
            "rank",
            "content_hash_id",
            "client_hash_id",
            "content_age_days",
            "search_volume",
            "content_type",
            "main_intent",
            "model_probability",
            "reason_code",
            "recommended_action"
        ]
    ].head(25)
)
```

```python
# ============================================================
# OPPORTUNITY DISTRIBUTION
# ============================================================

print("Validated opportunity by content type:")
print(
    validated_opportunities["content_type"]
    .value_counts()
)

print("\nValidated opportunity by intent:")
print(
    validated_opportunities["main_intent"]
    .value_counts()
)

print("\nValidated opportunity by recommended action:")
print(
    validated_opportunities["recommended_action"]
    .value_counts()
)

print("\nModel-only opportunity by content type:")
print(
    model_only_opportunities["content_type"]
    .value_counts()
)
```

```python
# ============================================================
# MODEL DISCOVERY QUEUE
# ============================================================

model_discovery_queue = model_only_opportunities.copy()

model_discovery_queue["discovery_score"] = (
    model_discovery_queue["model_probability"]
)

model_discovery_queue = (
    model_discovery_queue
    .sort_values(
        "discovery_score",
        ascending=False
    )
    .reset_index(drop=True)
)

model_discovery_queue["discovery_rank"] = (
    model_discovery_queue.index + 1
)

print("MODEL DISCOVERY QUEUE")
print(
    "Total model-only opportunities:",
    len(model_discovery_queue)
)

display(
    model_discovery_queue[
        [
            "discovery_rank",
            "content_hash_id",
            "client_hash_id",
            "content_age_days",
            "search_volume",
            "model_probability",
            "main_intent",
            "content_type",
            "recommended_action"
        ]
    ].head(25)
)
```

```python
# ============================================================
# FINAL BUSINESS OPPORTUNITY SCORE
# ============================================================

import numpy as np

final_opportunities = validated_opportunities.copy()

# Clean search volume
final_opportunities["search_volume_clean"] = (
    final_opportunities["search_volume"]
    .fillna(0)
    .clip(lower=0)
)

# Log transform search volume so extremely large volumes
# don't completely dominate the score
final_opportunities["volume_score"] = np.log1p(
    final_opportunities["search_volume_clean"]
)

# Normalize volume to 0-1
max_volume_score = final_opportunities["volume_score"].max()

if max_volume_score > 0:
    final_opportunities["volume_score_norm"] = (
        final_opportunities["volume_score"] /
        max_volume_score
    )
else:
    final_opportunities["volume_score_norm"] = 0

# Final business opportunity score
#
# Business-priority score:
# 60% model probability
# 40% search-demand strength

# ------------------------------------------------------------
# Business weighting
# ------------------------------------------------------------

MODEL_WEIGHT = 0.60
VOLUME_WEIGHT = 0.40

final_opportunities["business_opportunity_score"] = (
    MODEL_WEIGHT *
    final_opportunities["model_probability"]
    +
    VOLUME_WEIGHT *
    final_opportunities["volume_score_norm"]
)
# Rank
final_opportunities = (
    final_opportunities
    .sort_values(
        "business_opportunity_score",
        ascending=False
    )
    .reset_index(drop=True)
)

final_opportunities["final_rank"] = (
    final_opportunities.index + 1
)

print("FINAL BUSINESS OPPORTUNITY QUEUE")
print("Total opportunities:", len(final_opportunities))

display(
    final_opportunities[
        [
            "final_rank",
            "content_hash_id",
            "client_hash_id",
            "content_age_days",
            "search_volume",
            "model_probability",
            "volume_score_norm",
            "business_opportunity_score",
            "main_intent",
            "content_type",
            "recommended_action"
        ]
    ].head(25)
)
```

```python
# ============================================================
# TOP OPPORTUNITIES BY SEARCH DEMAND
# ============================================================

print("TOP 20 BY SEARCH VOLUME")

display(
    final_opportunities[
        [
            "content_hash_id",
            "client_hash_id",
            "search_volume",
            "model_probability",
            "business_opportunity_score",
            "main_intent",
            "content_type",
            "recommended_action"
        ]
    ]
    .sort_values("search_volume", ascending=False)
    .head(20)
)
```

```python
# ============================================================
# TOP OPPORTUNITIES BY MODEL PROBABILITY
# ============================================================

print("TOP 20 BY MODEL PROBABILITY")

display(
    final_opportunities[
        [
            "content_hash_id",
            "client_hash_id",
            "search_volume",
            "model_probability",
            "business_opportunity_score",
            "main_intent",
            "content_type",
            "recommended_action"
        ]
    ]
    .sort_values("model_probability", ascending=False)
    .head(20)
)
```

```python
# ============================================================
# FINAL OPPORTUNITY DISTRIBUTION
# ============================================================

print("========== FINAL OPPORTUNITY SUMMARY ==========")

print("Validated opportunities:", len(final_opportunities))

print("\nBy intent:")
display(
    final_opportunities["main_intent"]
    .value_counts()
    .rename_axis("main_intent")
    .reset_index(name="count")
)

print("\nBy content type:")
display(
    final_opportunities["content_type"]
    .value_counts()
    .rename_axis("content_type")
    .reset_index(name="count")
)

print("\nBy recommended action:")
display(
    final_opportunities["recommended_action"]
    .value_counts()
    .rename_axis("recommended_action")
    .reset_index(name="count")
)

print("\nSearch volume statistics:")
print(
    final_opportunities["search_volume"]
    .describe()
)
```
